```python
import matplotlib.pyplot as plt

# Seus dados
dados = [
    (6836, 5.00, 0.00),
    (7238, 4.85, 0.15),
    (7640, 4.66, 0.34),
    (8042, 4.47, 0.53),
    (8444, 4.30, 0.70),
    (8847, 4.13, 0.87),
    (9248, 3.96, 1.04),
    (9651, 3.81, 1.19),
    (10053, 3.66, 1.34),
    (10456, 3.51, 1.49),
    (10858, 3.38, 1.62),
    (11260, 3.25, 1.75),
    (11663, 3.11, 1.89),
    (12065, 2.99, 2.01),
    (12468, 2.87, 2.13),
    (12869, 2.76, 2.24),
    (13272, 2.65, 2.35),
    (13674, 2.55, 2.45),
    (14076, 2.45, 2.55),
    (14479, 2.35, 2.65),
    (14881, 2.26, 2.74)
]

# Separando os dados
x = [item[0] for item in dados]
y1 = [item[1] for item in dados]  # Tensão no resistor
y2 = [item[2] for item in dados]  # Tensão no capacitor


# --- Gráfico 1: Tensão no Capacitor (Vc) ---
plt.figure(figsize=(10, 6))
plt.plot(x, y2, 'b-o', label='Tensão no Capacitor (Vc)', linewidth=2, markersize=5)
plt.title('Curva de Carga do Capacitor (Vc)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Tempo (ms)', fontsize=12)
plt.ylabel('Tensão (V)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()


# --- Gráfico 2: Tensão no Resistor (Vr) ---
plt.figure(figsize=(10, 6))
plt.plot(x, y1, 'r-o', label='Tensão no Resistor (Vr)', linewidth=2, markersize=5)
plt.title('Curva de Descarga no Resistor (Vr)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Tempo (ms)', fontsize=12)
plt.ylabel('Tensão (V)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()


# --- Gráfico 3: Curvas Combinadas ---
plt.figure(figsize=(10, 6))
plt.plot(x, y1, 'r-o', label='Tensão no Resistor (Vr)', linewidth=2, markersize=5)
plt.plot(x, y2, 'b-o', label='Tensão no Capacitor (Vc)', linewidth=2, markersize=5)
plt.title('Comportamento de Carga e Descarga no Circuito RC', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Tempo (ms)', fontsize=12)
plt.ylabel('Tensão (V)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
